```python
from importlib.metadata import version, PackageNotFoundError
def v(p):
    try:
        return version(p)
    except PackageNotFoundError:
        return "not installed"
print("ipython-sql:", v("ipython-sql"))
print("sql (conflicting pkg):", v("sql"))
```

```python
import pathlib, sql

root = pathlib.Path(sql.__file__).parent
hits = []
for py in root.rglob("*.py"):
    try:
        txt = py.read_text(errors="ignore")
        if "MetaData(" in txt and "bind=" in txt:
            hits.append(str(py))
    except Exception:
        pass

print("Files using MetaData(..., bind=...):")
print("\n".join(hits) if hits else "None found")
```
