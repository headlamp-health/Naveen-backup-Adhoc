````python
# --- Reusable Postgres connection helpers (no installs needed) ---

from sqlalchemy.engine import URL
from sqlalchemy import create_engine

# >>> Put your creds here (or load from env/secret manager) <<<
PG_USER = "drugs_user"
PG_PASS = "drug"
PG_HOST = "beacon.cluster-cqtutorchzm4.us-east-2.rds.amazonaws.com"
PG_PORT = 5432
PG_DB   = "drugs"

def build_conn_url(sslmode: str = "require") -> str:
    """Return a SQLAlchemy URL string for Postgres (psycopg2 driver)."""
    url = URL.create(
        "postgresql+psycopg2",
        username=PG_USER,
        password=PG_PASS,     # URL.create auto-encodes special chars
        host=PG_HOST,
        port=PG_PORT,
        database=PG_DB,
        query={"sslmode": sslmode},
    )
    return str(url)

def get_engine(sslmode: str = "require"):
    """Return a SQLAlchemy Engine (use for pandas, writes, transactions)."""
    return create_engine(build_conn_url(sslmode), pool_pre_ping=True)

def connect_sql_magic(sslmode: str = "require"):
    """Connect the Jupyter %sql magic to Postgres using the current creds."""
    # Load extension if not already
    try:
        get_ipython().run_line_magic("load_ext", "sql")
    except Exception:
        pass
    conn_url = build_conn_url(sslmode)
    # Try both var syntaxes depending on extension (JupySQL vs ipython-sql)
    try:
        get_ipython().run_line_magic("sql", f"${'conn_url'}")   # JupySQL style
    except Exception:
        get_ipython().run_line_magic("sql", f"{conn_url}")      # ipython-sql style
    return conn_url

def get_completion(prompt, model=model_name):
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0, # this is the degree of randomness of the model's output
    )
    return response.choices[0].message.content


# --- 1) Prompt template (safe: no f-string braces) ---
TEMPLATE = Template(r"""
Role: You are a pharmacovigilance physician specializing in label intelligence. Extract ONLY items that are explicitly contraindicated from the provided text.

Input
Do the task briefed within triple backticks. Use ONLY the text inside:
```$text```

Scope & Definitions
- Include statements that use terms like “contraindicated”, “is/are contraindicated”, “must not be used”, “do not use”, “should not be administered” WHEN they appear in the Contraindications section text provided.
- Capture absolute prohibitions and their gating conditions (e.g., “within 14 days of MAOI use”).
- CI categories may include: hypersensitivity/allergy, drug class or specific drug co-administration, co-morbid conditions/physiologic states, lab/functional thresholds, procedures/devices, age groups, routes.
- Ignore Warnings/Precautions/Cautions unless they are explicitly inside the provided Contraindications text.
- Exclude negated forms (e.g., “not contraindicated”).

Extraction Rules
1) POSITIVE, VERBATIM: Pull exact phrases; do not paraphrase the core contraindicated term.
2) CONDITIONALS: Record timing/thresholds exactly (e.g., “within 14 days”, “QTc > 500 ms”).
3) DRUGS & CLASSES: Distinguish specific drug vs class (e.g., linezolid vs MAOIs).
4) HYPERSENSITIVITY: Capture as written (active substance, excipients, class cross-reactivity).
5) PROVENANCE: Provide character start–end offsets within the provided text when possible.
6) DEDUPLICATION: Merge identical concepts; keep the entry with the richest qualifiers.
7) NO ASSUMPTIONS: If a detail isn’t present (e.g., code, threshold), set null.
8) REPORTED CONDITIONS: For each “has been reported” statement, capture the condition term verbatim, the cue phrase, and product scope (this product vs. other products in same class).
9) SELF-MENTION FILTER (specific_drug only):
   Goal: remove self-mentions of the index drug, including salts/esters/abbreviations and formulation variants.
   Canonicalization steps (apply to BOTH the index drug + index_aliases and each extracted specific_drug):
     • lowercase
     • replace underscores/hyphens with space; remove punctuation/symbols; squeeze spaces
     • drop common salts/esters/forms (full & abbreviations):
       hydrochloride|hcl|hydrobromide|hbr|hemisuccinate|succinate|mesylate|tosylate|besylate|
       sulfate|phosphate|acetate|tartrate|maleate|fumarate|citrate|lactate|oxalate|
       sodium|na|potassium|k|magnesium|mg|calcium|ca|dihydrochloride|dimesylate|hemisulfate|palmitate
     • drop dosage/form/route modifiers:
       tablet|tabs|capsule|cap|solution|soln|suspension|syrup|injection|inj|patch|cream|gel|
       ointment|ophthalmic|otic|nasal|oral|topical|iv|im|sc|po|er|xr|sr|cr|la|ir|cd
     • drop strengths/units (e.g., “10 mg”, “5 mL”, percent)
   Comparison:
     • Build a canonical string for the index_drug and each index_alias.
     • Build a canonical string for each extracted specific_drug.
     • If an extracted specific_drug canonical equals ANY canonical index term, EXCLUDE that item from outputs.
   Examples:
     • “methylphenidate_hydrochloride 10 mg tabs” → canonical “methylphenidate” ⇒ matches index “methylphenidate” ⇒ exclude.
     • “fluoxetine HCl oral solution” → canonical “fluoxetine” ⇒ matches index “fluoxetine” ⇒ exclude.
     • “paliperidone palmitate injection” → canonical “paliperidone” ⇒ matches index “paliperidone” ⇒ exclude.
   Note: This filter applies ONLY to category "specific_drug". Do NOT remove entries categorized as "drug_class".



Output (JSON only)
{
  "contraindications": [
    {
      "category": "drug_class" | "specific_drug" | "condition" | "physiologic_state" | "lab_threshold" | "procedure_device" | "allergy_hypersensitivity" | "age_group" | "route",
      "term_verbatim": "<exact phrase from text>",
      "qualifiers": {
        "timing": null | "<e.g., within 14 days>",
        "threshold": null | "<e.g., CrCl < 30 mL/min or QTc > 500 ms>",
        "population": null | "<e.g., pregnancy, pediatric, severe hepatic impairment>",
        "route": null | "<if route-specific>"
      },
      "drug_normalization": {
        "specific_drug_name": null | "<as written>",
        "drug_class_name": null | "<as written>"
      },
      "codes": {
        "SNOMED": null | "<if code is present in text>",
        "ICD10": null | "<if code is present>",
        "ATC": null | "<if code is present>"
      },
      "provenance": {
        "section_header": "Contraindications",
        "char_span": [start_index, end_index]
      },
      "notes": null | "<alternate phrasing or cross-reference>"
    }
  ],
  "reported_conditions": [
      {
        "condition_term_verbatim": "<e.g., Angioedema>",
        "cue_phrase": "<e.g., has been reported / postmarketing reports of>",
        "product_scope": "this_product" | "other_products_same_class" | "class_unspecified",
        "related_products_text": null | "<e.g., other methylphenidate products>",
        "qualifiers": {
          "population": null | "<e.g., pediatric>",
          "timing": null | "<if any>",
          "severity": null | "<e.g., serious, life-threatening, rare>"
        },
        "provenance": {
          "section_header": "<as provided or nearby heading>",
          "char_span": [start_index, end_index]
        },
        "notes": null | "<alternate phrasing>"
      }
     ], 
  "meta": {
    "source_sections_seen": ["Contraindications"],
    "extraction_confidence": 0.0-1.0
  }
}

Post-processing
- Remove any entry not supported by explicit text in ```$text```.
- If the same concept appears multiple times, keep the most complete entry and record alternates in notes.
- Sort entries by term_verbatim (A→Z).
- Return ONLY the JSON.
""")

# --- 2) Helper: robust JSON parsing (tolerates extra text around JSON) ---
def parse_json_response(raw: str):
    # Fast path
    try:
        return json.loads(raw)
    except Exception:
        pass
    # Try to extract the largest {...} block
    try:
        start = raw.find('{')
        end = raw.rfind('}')
        if start != -1 and end != -1 and end > start:
            candidate = raw[start:end+1]
            return json.loads(candidate)
    except Exception:
        pass
    return None  # could not parse

def play_music(): 
    # List of royalty-free sample audio URLs
    music_urls = [
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-1.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-2.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-3.mp3"
    ]
    
    # Pick one at random
    selected_url = random.choice(music_urls)
    
    # Trigger point in your logic
    print("🎵 Trigger reached, playing random music from web...")
    
    # Play
    display(Audio(url=selected_url, autoplay=True))

def _clean(s: str):
    if not isinstance(s, str): return None
    s = s.strip()
    s = re.sub(r"\s+", " ", s)
    return s or None

def _canon_key_class(s: str):
    if not isinstance(s, str): return None
    s = s.lower()
    s = re.sub(r"[^\w\s]", "", s)      # rm punctuation
    s = re.sub(r"\s+", " ", s).strip()
    return s or None

def _canon_key_drug(s: str):
    if not isinstance(s, str): return None
    s = s.lower().replace("_", " ").replace("-", " ")
    # remove salts/esters/ions, form modifiers, and strengths
    s = re.sub(r"\b(hcl|hydrochloride|hbr|hydrobromide|mesylate|tosylate|besylate|"
               r"sulfate|phosphate|acetate|tartrate|maleate|fumarate|citrate|lactate|oxalate|"
               r"sodium|na|potassium|k|magnesium|mg|calcium|ca|palmitate|succinate|hemisulfate|"
               r"dihydrochloride|dimesylate)\b", "", s)
    s = re.sub(r"\b(er|xr|sr|cr|ir|la|cd|tab(?:s)?|caps?(?:ules?)?|solution|soln|susp(?:ension)?|"
               r"syrup|injection|inj|patch|cream|gel|ointment|ophthalmic|otic|nasal|oral|topical|iv|im|sc|po)\b", "", s)
    s = re.sub(r"\b\d+(\.\d+)?\s*(mg|mcg|g|ml|%)\b", "", s)
    s = re.sub(r"[^\w\s]", "", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s or None

def _dedupe_preserve_order(values, key_fn):
    if not isinstance(values, list): 
        return []
    out, seen = [], set()
    for v in values:
        v_clean = _clean(v)
        if not v_clean: 
            continue
        k = key_fn(v_clean)
        if not k or k in seen:
            continue
        seen.add(k)
        out.append(v_clean)
    return out
    
def extract_classes_and_drugs(parsed):
    # maps canonical_key -> display_value
    class_map, drug_map = {}, {}

    if not isinstance(parsed, dict):
        return [], []

    for item in parsed.get("contraindications", []):
        if not isinstance(item, dict):
            continue

        raw_cat = (
            item.get("category")
            or item.get("Category")
            or item.get("specific_category")
            or item.get("Specific Category")
        )
        cat = str(raw_cat).strip().lower() if raw_cat is not None else None
        if cat not in {"specific_drug", "drug_class"}:
            continue

        dn = item.get("drug_normalization") or {}

        if cat == "drug_class":
            val = dn.get("drug_class_name") or item.get("term_verbatim")
            val = _clean(val)  # your existing cleaner
            key = _canon_key(val)
            if val and key and key not in class_map:
                class_map[key] = val

        elif cat == "specific_drug":
            val = dn.get("specific_drug_name") or item.get("term_verbatim")
            val = _clean(val)  # your existing cleaner
            key = _canon_key_drug(val)
            if val and key and key not in drug_map:
                drug_map[key] = val

    # return sorted, unique lists (sorted by casefold for stable order)
    classes = sorted(class_map.values(), key=str.casefold)
    drugs   = sorted(drug_map.values(), key=str.casefold)
    return classes, drugs

def canonicalize(name: str):
    if not isinstance(name, str):
        return None
    # lowercase, strip punctuation-y clutter, squeeze spaces
    s = name.lower().strip()
    s = re.sub(r"[\u2018\u2019\u201c\u201d'\"®™©]", "", s)        # quotes/symbols
    s = re.sub(r"\b(hcl|hydrochloride|sodium|phosphate)\b", "", s) # common salts (tune as needed)
    s = re.sub(r"\b(tab(?:let)?s?|caps?(?:ule)?s?|susp(?:ension)?|inj(?:ection)?|soln|sr|er|xr|cr|cd)\b", "", s)
    s = re.sub(r"\b\d+(\.\d+)?\s*(mg|mcg|g|ml)\b", "", s)         # strengths
    s = re.sub(r"\s*[-–]\s*", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s or None

def build_synonyms(row):
    """
    Build a set of canonical synonyms for the index drug in this row.
    Populate from whatever columns you have (examples shown).
    """
    cand = set()
    for col in ["index_drug", "generic_name", "active_ingredient", "brand_name", "brand_names"]:
        if col in row and row[col] is not None:
            val = row[col]
            if isinstance(val, (list, tuple, set)):
                for x in val:
                    c = canonicalize(str(x))
                    if c: cand.add(c)
            else:
                c = canonicalize(str(val))
                if c: cand.add(c)
    return cand

def filter_self_mentions(classes_list, drugs_list, row):
    """
    Remove specific drugs that match the index drug (or its synonyms).
    """
    if not isinstance(drugs_list, list):
        return classes_list, drugs_list
    syns = build_synonyms(row)
    if not syns:
        return classes_list, drugs_list
    filtered = []
    for d in drugs_list:
        c = canonicalize(d)
        if c and c in syns:
            continue  # drop self-mention
        filtered.append(d)
    return classes_list, filtered

def coerce_to_list(x):
    if isinstance(x, list): return x
    if x is None: return []
    s = str(x).strip()
    if not s: return []
    # looks like a Python list: "['a','b']" -> treat as a single string fallback; we'll split later
    return [s]

# 2) Normalize a single token
def _norm(s: str) -> str:
    s = unicodedata.normalize("NFKC", s or "")
    s = s.replace("\u200b","").replace("\ufeff","")
    s = " ".join(s.strip().split())
    return s

# 3) Split any element that contains multiple values ("a, b / c and d")
TOKEN_SPLIT_RE = re.compile(r"\s*(?:,|;|/|\band\b|\bor\b|\band/or\b)\s*", flags=re.IGNORECASE)

def tokenize_flatten(values):
    """Take a mixed list like ['a', 'a, b', 'c'] -> ['a','a','b','c']."""
    out = []
    for v in (values or []):
        if v is None: 
            continue
        for t in TOKEN_SPLIT_RE.split(str(v)):
            t = _norm(t)
            if t:
                out.append(t)
    return out

# 4) Dedupe while preserving order (case-insensitive)
def dedupe_keep_order(values):
    out, seen = [], set()
    for v in values:
        key = _norm(v).lower()
        if key and key not in seen:
            seen.add(key)
            out.append(_norm(v))
    return out
````

````python
df[["ci_drug_classes", "ci_specific_drugs"]] = df.apply(
    lambda row: pd.Series(
        filter_self_mentions(row.get("ci_drug_classes", []), row.get("ci_specific_drugs", []), row)
    ),
    axis=1
)

# Optional counts after filtering:
df["ci_specific_drug_count"] = df["ci_specific_drugs"].map(lambda x: len(x) if isinstance(x, list) else 0)
df["ci_drug_classes"]   = df["ci_drug_classes"].apply(lambda lst: _dedupe_preserve_order(lst, _canon_key_class))
df["ci_specific_drugs"] = df["ci_specific_drugs"].apply(lambda lst: _dedupe_preserve_order(lst, _canon_key_drug))
````

````python
df["ci_specific_drugs"] = df["ci_specific_drugs"].map(coerce_to_list)
df["ci_drug_classes"]   = df["ci_drug_classes"].map(coerce_to_list)

# Tokenize inside each element, then normalize + dedupe
df["ci_specific_drugs"] = (
    df["ci_specific_drugs"]
      .map(tokenize_flatten)
      .map(dedupe_keep_order)
)

df["ci_drug_classes"] = (
    df["ci_drug_classes"]
      .map(tokenize_flatten)
      .map(dedupe_keep_order)
)
````
